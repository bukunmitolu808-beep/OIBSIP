```python
import pandas as pd

df = pd.read_csv("Messy_Employee_dataset.csv")
print("Dataset loaded!")
print("Shape:", df.shape)
df.head(10)
```

```python

# Column names and data types
df.info()
```

```python
# Summary statistics for numerical columns
df.describe()
```

```python
# Check for missing values
df.isna().sum()
```

```python
# Check for duplicate rows
duplicate_rows = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_rows)
```

```python
# Convert Join_Date to proper datetime format
df["Join_Date"] = pd.to_datetime(df["Join_Date"], errors="coerce")
```

```python
# Fill missing Age with median age
median_age = df["Age"].median()
df["Age"] = df["Age"].fillna(median_age)

# Optionally convert Age to integer (if appropriate)
df["Age"] = df["Age"].astype(int)
```

```python
# Fill missing Salary with median salary
median_salary = df["Salary"].median()
df["Salary"] = df["Salary"].fillna(median_salary)
```

```python
# Drop the Phone column because values are invalid
df = df.drop(columns=["Phone"])
```

```python
# Replace invalid phone numbers with NaN
df["Phone"] = pd.NA
```

```python
# Standardise Status
df["Status"] = df["Status"].astype(str).str.strip().str.title()

# Standardise Performance_Score
df["Performance_Score"] = df["Performance_Score"].astype(str).str.strip().str.title()
```

```python
df["Department_Region"] = df["Department_Region"].astype(str).str.strip().str.title()
```

```python
# Re-check missing values
print("Missing values after cleaning:\n", df.isna().sum())

# Re-check data types
df.info()

# Quick look at first rows
df.head(10)
```

```python
# Export cleaned dataset
df.to_csv("cleaned_employee_data.csv", index=False)
print("Cleaned data saved as cleaned_employee_data.csv")
```
